```python
import numpy as np 
np.random.seed(42)
X = np.random.randn(5,4)
X
```

```python
from selfattention import SelfAttention
from normalization import LayerNorm
import numpy as np
np.random.seed(42)


layer_norm = LayerNorm(4)
self_attention = SelfAttention(4, 2, 2)

# LayerNorm
X_norm = layer_norm.normalize(X)

# Q, K, V
Q, K, V = self_attention.project(X_norm)

# Attention scores
attention_scores = self_attention.attention_scores(Q, K)

# Process each head
scaled_scores = []
masked_scores = []
attention_weights = []

for i in range(self_attention.num_heads):
    scaled = self_attention.scale_scores(attention_scores[i])
    masked = self_attention.causal_mask(scaled)
    weights = self_attention.softmax(masked)
    scaled_scores.append(scaled)
    masked_scores.append(masked)
    attention_weights.append(weights)

# Output for all heads
outputs = self_attention.output(attention_weights, V)

# Concatenate heads
concatenated = self_attention.concatenate(outputs)

# Output projection
mha_output = self_attention.output_projection(concatenated)

# Residual
residual_output = X + mha_output

print("Original X shape:", X.shape)
print("LayerNorm output shape:", X_norm.shape)
print("MHA output shape:", mha_output.shape)
print("Residual output shape:", residual_output.shape)

print("\nResidual output:")
print(residual_output)
```
